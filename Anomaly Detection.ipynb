**Step 2: Concepts Reminder**

Anomaly → A data point that deviates significantly from the norm.

Outlier → Similar to anomaly, but often statistical noise.

Z-Score → Measures how far a point is from the mean.

Isolation Forest → ML algorithm that isolates anomalies by random partitioning.

Applications → Fraud detection, medical monitoring, cybersecurity.

**Step 3: Dataset Example**

Let’s assume a dataset of transaction amounts:

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'Transaction_ID': range(1, 11),
    'Amount': [50, 52, 49, 51, 500, 48, 53, 47, 49, 1000]
})
data
```

Here, most transactions are ~50, but 500 and 1000 look suspicious.

**Step 4: Statistical Detection (Z-Score)**

```python
from scipy import stats

z_scores = np.abs(stats.zscore(data['Amount']))
data['Z_Score'] = z_scores
data['Anomaly'] = data['Z_Score'] > 3
data
```

Transactions with Z-Score > 3 are anomalies.

Detected anomalies: 500, 1000.

**Step 5: Machine Learning Detection (Isolation Forest)**

```python
from sklearn.ensemble import IsolationForest

iso = IsolationForest(contamination=0.2, random_state=42)
data['IF_Anomaly'] = iso.fit_predict(data[['Amount']])
data
```

IF_Anomaly = -1 → anomaly.

Isolation Forest also flags 500 and 1000 as anomalies.

| **Method** | **Strength** | **Limitation** |
| --- | --- | --- |
| **Z-Score** | Simple, fast | Assumes normal distribution |
| **Isolation Forest** | Works for complex data | Requires tuning contamination parameter |
